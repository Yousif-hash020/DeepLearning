```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
document = [
    "I love pizza",
    "pizza is the best",
    "i love pasta",
    "pasta is the best",
    "i love pizza pasta"
]
```

```python
vectorize = CountVectorizer()
```

```python
X =vectorize.fit_transform(document)
```

```python
print("vocabulary", vectorize.get_feature_names_out())
print("\n BOW matix :\n", X.toarray())
```
